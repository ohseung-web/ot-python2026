```python
# 1. 파일을 읽기 모드로 열어 모든 줄을 리스트로 가져옵니다.
with open('greenData/text_data.txt', 'r', encoding='utf-8') as f:
    lines = f.readlines()

# 2. 삭제하고 싶은 인덱스를 선택해 삭제합니다.
# 예: 첫 번째 줄 삭제 시 lines.pop(0) 또는 del lines[0]
target_index = 0 
if len(lines) > target_index:
    lines.pop(target_index)

# 3. 삭제된 리스트 내용을 파일에 다시 씁니다 ('w' 모드는 기존 내용을 덮어씁니다).
with open('greenData/text_data.txt', 'w', encoding='utf-8') as f:
    f.writelines(lines)

print(lines)    
```

```python
import csv

headers = ['ID', 'Score']
with open('greenData/result.csv', 'w', encoding='utf-8-sig', newline='') as f:
    writer = csv.DictWriter(f, fieldnames=headers)
    writer.writeheader() # 헤더 기록
    writer.writerow({'ID': 'user01', 'Score': 95})
    writer.writerow({'ID': 'user02', 'Score': 88})

# 2. 파일 읽기 부분 (DictReader로 수정)
with open('greenData/result.csv', 'r', encoding='utf-8-sig') as f:
    # csv.reader 대신 csv.DictReader 사용
    reader = csv.DictReader(f)
    
    for row in reader:
        print(dict(row)) # dict()로 감싸주면 깔끔하게 딕셔너리 형태로 출력됩니다.
```

```python
# member_data.xml 파일을 생성하는 코드입니다.
xml_content = """<?xml version="1.0" encoding="UTF-8"?>
<members>
    <member>
        <name>김철수</name>
        <age>25</age>
        <city>서울</city>
    </member>
    <member>
        <name>이영희</name>
        <age>30</age>
        <city>부산</city>
    </member>
    <member>
        <name>박민수</name>
        <age>22</age>
        <city>대전</city>
    </member>
</members>
"""

# 파일 저장
with open('greenData/member_data.xml', 'w', encoding='utf-8') as f:
    f.write(xml_content)

print("member_data.xml 파일이 성공적으로 생성되었습니다!")
```

```python
from bs4 import BeautifulSoup

# 1. 파일 읽기
with open('greenData/member_data.xml', 'r', encoding='utf-8-sig') as f:
    # 2. 파싱 (XML 전용 파서 사용)
    soup = BeautifulSoup(f, 'xml')

# 3. 데이터 추출 (예: 'member' 태그 선택)
members = soup.select('member')

for m in members:
    name = m.select_one('name').text
    age = m.select_one('age').text
    print(f"이름: {name}, 나이: {age}")
```

```python
import requests
from bs4 import BeautifulSoup

keyword = '파이썬'
api_url = f'http://suggestqueries.google.com/complete/search?output=toolbar&q={keyword}'

# API 호출 및 결과 받기
response = requests.get(api_url)
# 객체 내부의 content를 XML로 파싱
soup = BeautifulSoup(response.content, 'xml')

# 'suggestion' 태그들 선택
items = soup.select('suggestion')

print(f"[{keyword}] 관련 검색어 리스트:")
for item in items:
    # 속성값 추출 시 dict처럼 ['key'] 사용
    print("-", item['data'])
```

```python
from bs4 import BeautifulSoup

# 방금 만든 파일 읽기
with open('greenData/member_data.xml', 'r', encoding='utf-8') as f:
    soup = BeautifulSoup(f, 'xml')

# 모든 member 태그 찾기
members = soup.select('member')

for m in members:
    # 각 태그 안의 텍스트 추출
    name = m.select_one('name').text
    age = m.select_one('age').text
    city = m.select_one('city').text
    print(f"이름: {name} | 나이: {age} | 지역: {city}")
```

```python
import json

# 1. JSON 문자열 -> 사전 (Decoding)
json_str = '{"name": "Alice", "role": "admin", "level": 5}'
data = json.loads(json_str)
print(data['name'], type(data))

# 2. 사전 -> JSON 문자열 (Encoding)
user_dict = {"id": 7, "status": "online"}
# indent 옵션을 사용하면 가독성 있게 들여쓰기 처리됨
readable_json = json.dumps(user_dict, indent=4)
print(readable_json)
```

```python
import json

# 데이터 준비
project_data = {
    "title": "Data Analysis",
    "tags": ["Pandas", "JSON", "Automation"],
    "status": "In Progress"
}

# 1. 파일 쓰기 (Save)
with open('greenData/project.json', 'w', encoding='utf-8') as f:
    json.dump(project_data, f, indent=4)

# 2. 파일 읽기 (Read)
with open('greenData/project.json', 'r', encoding='utf-8') as f:
    loaded_data = json.load(f)
    print(loaded_data.keys())
```

```python
import json

# 파일 읽기 테스트
try:
    with open('greenData/category_data.json', 'r', encoding='utf-8') as f:
        data = json.load(f)
        
    print(f"데이터 종류: {data['kind']}")
    print("-" * 30)
    
    for item in data['items']:
        cat_id = item['id']
        title = item['snippet']['title']
        print(f"카테고리 번호: {cat_id} | 이름: {title}")
        
except FileNotFoundError:
    print("파일이 없습니다. 파일명을 확인해 주세요!")
```

```python
import json

# 파일 로드 (앞서 만든 category_data.json 활용)
with open('category_data.json', 'r', encoding='utf-8') as f:
    data = json.load(f)

# 리스트 컴프리헨션으로 title만 추출
titles = [item['snippet']['title'] for item in data.get('items', [])]
print(f"추출된 카테고리: {titles}")
```

```python
# ID가 20 이상인 카테고리만 필터링
filtered_items = [item for item in data['items'] if int(item['id']) >= 20]
for res in filtered_items:
    print(f"필터링 결과: {res['snippet']['title']} (ID:{res['id']})")
```

```python
import json
import csv

# 1. JSON 읽기
with open('category_data.json', 'r', encoding='utf-8') as f:
    json_data = json.load(f)

# 2. CSV 쓰기
with open('converted_data.csv', 'w', newline='', encoding='utf-8-sig') as f:
    writer = csv.writer(f)
    # 헤더 작성
    writer.writerow(['ID', 'Title', 'ChannelID'])
    
    # 데이터 작성
    for item in json_data['items']:
        writer.writerow([
            item['id'], 
            item['snippet']['title'], 
            item['snippet']['channelId']
        ])
print("변환이 완료되었습니다.")
```
