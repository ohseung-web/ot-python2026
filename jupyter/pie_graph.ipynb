```python
import pandas as pd
import matplotlib.pyplot as plt

# ---------------------------
# 1. 데이터 생성
# ---------------------------
df1 = pd.DataFrame({
    "지역": ["서울", "부산", "대구"],
    "총인구": [1000000, 300000, 200000]
})

df2 = pd.DataFrame({
    "지역": ["서울", "부산", "인천"],
    "병원수": [150, 80, 60]
})

# ---------------------------
# 2. 데이터 병합 (지역 기준)
# how='outer' → 모든 지역 포함
# ---------------------------
merged= pd.merge(df1, df2, on="지역", how="outer")

# ---------------------------
# 3. 결측치 처리
# 총인구가 없는 값 → 평균으로 채우기
# ---------------------------
merged["총인구"] = merged["총인구"].fillna(merged["총인구"].mean())
merged['병원수'] = merged['병원수'].fillna(merged['병원수'].mean())
# 결과 확인
print(merged)

# ---------------------------
# 4. 한글 깨짐 방지 설정
# ---------------------------
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

plt.figure(figsize=(6, 6))

plt.pie(
    merged["병원수"],
    labels=merged["지역"],
    autopct='%1.1f%%',
    startangle=90  # 시작 위치 (위쪽부터 시작)
)

plt.title("지역별 병원 수 비율")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# ---------------------------
# 1. 데이터 생성
# ---------------------------
df1 = pd.DataFrame({
    "지역": ["서울", "부산", "대구"],
    "총인구": [1000000, 300000, 200000]
})

df2 = pd.DataFrame({
    "지역": ["서울", "부산", "인천"],
    "병원수": [150, 80, 60]
})

# ---------------------------
# 2. 데이터 병합
# ---------------------------
merged = pd.merge(df1, df2, on="지역", how="outer")

# ---------------------------
# 3. 결측치 처리
# ---------------------------
merged["총인구"] = merged["총인구"].fillna(merged["총인구"].mean())
merged["병원수"] = merged["병원수"].fillna(merged["병원수"].mean())

print(merged)

# ---------------------------
# 4. Seaborn 스타일 + 한글 설정
# ---------------------------
sns.set_style('whitegrid')
sns.set_palette('deep')

plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

# ---------------------------
# 5. 그래프 크기
# ---------------------------
plt.figure(figsize=(6, 6))

colors = ['#ff9999', '#66b3ff', '#99ff99', '#ffcc99']

# explode 설정
# 각 조각을 중심에서 얼마나 떨어뜨릴지 지정하는 값
# 리스트 길이는 데이터 개수와 동일해야 함
# [0.05, 0.05, 0.05, 0.05] 와 동일한 의미
# → 모든 조각을 동일하게 살짝 바깥으로 이동 (강조 효과)
explode = [0.05] * len(merged)

plt.pie(
    merged["병원수"],          # 데이터 값 (각 조각의 크기)
    labels=merged["지역"],     # 각 조각의 이름
    autopct='%1.1f%%',        # 퍼센트 표시 (소수점 1자리)
    startangle=90,            # 시작 위치 (위쪽부터 시작)
    colors=colors,            # 각 조각 색상
    explode=explode,          # 조각을 바깥으로 얼마나 띄울지 설정
    wedgeprops={'edgecolor': 'white'},  # 조각 사이 흰색 경계선

    textprops={
        'fontsize': 12,      # 글자 크기
        'color': 'black'     # 글자 색
    }
)

# ---------------------------
# 8. 제목
# ---------------------------
plt.title("지역별 병원 수 비율")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# ---------------------------
# 1. 데이터 생성
# ---------------------------
df1 = pd.DataFrame({
    "지역": ["서울", "부산", "대구"],
    "총인구": [1000000, 300000, 200000]
})

df2 = pd.DataFrame({
    "지역": ["서울", "부산", "인천"],
    "병원수": [150, 80, 60]
})

# ---------------------------
# 2. 데이터 병합
# ---------------------------
merged = pd.merge(df1, df2, on="지역", how="outer")

# ---------------------------
# 3. 결측치 처리
# ---------------------------
merged["총인구"] = merged["총인구"].fillna(merged["총인구"].mean())
merged["병원수"] = merged["병원수"].fillna(merged["병원수"].mean())

print(merged)


sns.set_style('whitegrid')

plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

plt.figure(figsize=(6, 6))

colors = sns.color_palette('pastel')

# ---------------------------
# 1. 파이(도넛) 그래프 생성
# ---------------------------
result = plt.pie(
    merged["병원수"],
    labels=merged["지역"],
    autopct='%1.1f%%',
    startangle=90,
    colors=colors,
    
    pctdistance=0.75,   #퍼센트를 도넛 안쪽으로 이동
    
    wedgeprops={
        'edgecolor': 'white',
        'width': 0.4
    }
)

# ---------------------------
# 2. 결과 나누기 
# ---------------------------
wedges = result[0]       # 조각 (거의 안 씀)
label_texts = result[1]  # 지역 이름 텍스트
percent_texts = result[2]  # 퍼센트 텍스트

# ---------------------------
# 3. 지역 이름 글자 크기 조절
# ---------------------------
for t in label_texts:
    t.set_fontsize(11)

# ---------------------------
# 4. 퍼센트 글자 꾸미기
# ---------------------------
for t in percent_texts:
    t.set_color('black')   # 검은색으로 변경
    t.set_fontsize(10)
# ---------------------------
# 9. 제목
# ---------------------------
plt.title("지역별 병원 수 비율 (도넛형)")

plt.tight_layout()
plt.show()
```
