```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

# ---------------------------
# 1. 데이터 생성
# ---------------------------
data = pd.DataFrame({
    "수학": [80, 90, 70],
    "영어": [85, 60, 75],
    "과학": [90, 88, 95]
}, index=["김학생", "이학생", "박학생"])


# ---------------------------
# 2. 히트맵
# ---------------------------
sns.heatmap(data)

plt.title("학생 성적 히트맵")
plt.show()
```
