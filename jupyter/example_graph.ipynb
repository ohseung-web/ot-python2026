```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.DataFrame({
    "과목": ["Java", "Python", "React", "SQL"],
    "점수": [85, 90, 75, 80]
})

# ---------------------------
# 4. 한글 깨짐 방지 설정
# ---------------------------
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False


bars = plt.bar(df["과목"], df["점수"])

# 막대 위 값 표시
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, height, str(height),
             ha='center', va='bottom')

plt.title("과목별 점수")
plt.xlabel("과목")
plt.ylabel("점수")

plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.DataFrame({
    "이름": ["김학생", "이학생", "박학생", "최학생"],
    "공부시간": [3, 5, 2, 4]
})

plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

plt.barh(df["이름"], df["공부시간"], height=0.5)

plt.title("학생별 공부시간")
plt.xlabel("공부시간")
plt.ylabel("이름")

plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.DataFrame({
    "월": ["1월", "2월", "3월", "4월"],
    "매출": [100, 150, 130, 170]
})

plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

x = df["월"]
y = df["매출"]

plt.plot(x, y, marker='o')  # 선그래프
plt.fill_between(x, y, alpha=0.3)  # 영역 채우기

plt.title("월별 매출 변화")
plt.xlabel("월")
plt.ylabel("매출")

plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.DataFrame({
    "카테고리": ["식비", "교통", "쇼핑", "기타"],
    "금액": [300000, 100000, 200000, 150000]
})

plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

plt.pie(
    df["금액"],
    labels=df["카테고리"],
    autopct='%1.1f%%',
    startangle=90
)

plt.title("지출 비율")

plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.DataFrame({
    "OS": ["Windows", "Mac", "Linux"],
    "사용자수": [70, 20, 10]
})

plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

plt.pie(
    df["사용자수"],
    labels=df["OS"],
    autopct='%1.1f%%',
    startangle=90
)

# 도넛 핵심 (가운데 원)
centre_circle = plt.Circle((0, 0), 0.6, fc='white')
plt.gca().add_artist(centre_circle)

plt.title("OS 점유율")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# ---------------------------
# 1. 데이터 생성
# ---------------------------
df1 = pd.DataFrame({
    "부서": ["개발팀", "영업팀", "인사팀"],
    "기본급": [300, 250, 200]
})

df2 = pd.DataFrame({
    "부서": ["개발팀", "영업팀", "마케팅팀"],
    "성과급": [100, 80, 70]
})

# ---------------------------
# 2. 데이터 병합
# ---------------------------
merged = pd.merge(df1, df2, on="부서", how="outer")

# ---------------------------
# 3. 결측치 처리 (평균)
# ---------------------------
merged["기본급"] = merged["기본급"].fillna(merged["기본급"].mean())
merged["성과급"] = merged["성과급"].fillna(merged["성과급"].mean())

print(merged)

plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

# ---------------------------
# 4. 그래프 생성
# ---------------------------
plt.figure(figsize=(8, 5))

# 기본급 (아래)
plt.bar(merged["부서"], merged["기본급"], label="기본급")

# 성과급 (위에 누적)
plt.bar(
    merged["부서"],
    merged["성과급"],
    bottom=merged["기본급"],  # 핵심
    label="성과급"
)

# ---------------------------
# 5. 그래프 꾸미기
# ---------------------------
plt.title("부서별 급여 구조")
plt.xlabel("부서")
plt.ylabel("금액")

plt.legend()  # 범례

# ---------------------------
# 6. 저장
# ---------------------------
plt.savefig("salary_chart.png", dpi=300)

plt.show()
```
