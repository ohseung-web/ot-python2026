```python
import pandas as pd

df = pd.read_csv("greenData/pm10.csv", encoding="cp949")

# 컬럼 공백 제거 (안전)
df.columns = df.columns.str.strip()

# 결측치 처리
df["미세먼지(PM10)"] = df["미세먼지(PM10)"].fillna(df["미세먼지(PM10)"].mean())

# 지역별 평균 (구분 컬럼 사용)
grouped = df.groupby("구분")["미세먼지(PM10)"].mean()

# 정렬
sorted_df = grouped.sort_values(ascending=False)

print(sorted_df.head(10))
```

```python
import pandas as pd

df = pd.read_csv("greenData/accident.csv",encoding="cp949")


# 날짜 변환
df["사고일자"] = pd.to_datetime(df["사고일자"])

# 연도 컬럼
df["연도"] = df["사고일자"].dt.year

# 그룹화
grouped = df.groupby("연도").size()

# 정렬
sorted_df = grouped.sort_values(ascending=False)

print(sorted_df)

# 최대 연도
print("최대 사고 연도:", sorted_df.idxmax())

# 전체 합
print("총 사고 건수:", grouped.sum())
```

```python
import pandas as pd

# 문제 1
df = pd.read_csv("greenData/population.csv", encoding="cp949")

# 문제 2
df["총인구수"] = df["남자인구수"] + df["여자인구수"]
grouped_population = df.groupby(["시도명", "시군구명"])["총인구수"].sum()

# 문제 3
sorted_population = grouped_population.sort_values(ascending=False)
print("상위 10개 시군구별 총 인구수:")
print(sorted_population.head(10))

# 문제 4
grouped_household = df.groupby(["시도명", "시군구명"])["세대수"].mean()
print("\n시군구별 평균 세대수:")
print(grouped_household)

# 문제 5
max_population_region = sorted_population.idxmax()
max_population_value = sorted_population.max()
print(f"\n가장 인구가 많은 시군구는 {max_population_region}이며, 인구수는 {max_population_value}명입니다.")
```

```python
import pandas as pd

df = pd.read_csv("greenData/hospital.csv",encoding="cp949")

# 지역별 개수
grouped = df.groupby("지역").size()

# 정렬
sorted_df = grouped.sort_values(ascending=False)

print(sorted_df.head(10))

# 평균
print("평균:", grouped.mean())

# 최대
print("최대 지역:", sorted_df.idxmax())
```

```python
import pandas as pd

# 문제 1
df = pd.read_csv("greenData/hospital.csv",encoding="cp949")

# 문제 2
# 소속기관별 병원 수 계산 (만약 소속기관이 하나라면 협력병원명별 중복 제거 후 개수)
hospital_counts = df.groupby("소속기관")["협력병원명"].nunique()

# 문제 3
sorted_counts = hospital_counts.sort_values(ascending=False)
print("상위 10개 소속기관별 병원 수:")
print(sorted_counts.head(10))

# 문제 4
mean_hospital_count = hospital_counts.mean()
print(f"\n평균 병원 수: {mean_hospital_count}")

# 문제 5
max_hospital_institution = hospital_counts.idxmax()
max_hospital_value = hospital_counts.max()
print(f"\n가장 병원이 많은 소속기관은 '{max_hospital_institution}'이며, 병원 수는 {max_hospital_value}개입니다.")
```

```python
import pandas as pd

df = pd.read_csv("cafe.csv")

# 지역별 개수
grouped = df.groupby("지역").size()

# 조건 필터링
filtered = grouped[grouped >= 100]

# 정렬
sorted_df = filtered.sort_values(ascending=False)

print(sorted_df.head(10))

# 총합
print("총 카페 수:", grouped.sum())

# 최대 지역
print("최대 지역:", sorted_df.idxmax())
```
